# 1) Histogramme

```python
import matplotlib.pyplot as plt
import seaborn as sns

notes = [10,11,12,13,14]
plt.hist(notes, bins=20)
plt.xlabel("Notes")
plt.ylabel("Fréquence")
plt.title("Distribution des notes")
plt.show()
```

# 2= Boxplot

```python
import matplotlib.pyplot as plt
import seaborn as sns

notes = [10,11,12,13,14]
plt.boxplot(notes)
plt.ylabel("Notes")
plt.title("Boxplot des notes")
plt.show()
```

# 3) Nuage de points

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

x = np.random.randint(0,21,size=40)
y = np.random.randint(0,21,size=40)
plt.scatter(x, y, color = "red", s = 100, facecolor = "red", edgecolor = "blue")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Relation X / Y")
plt.show()
```
