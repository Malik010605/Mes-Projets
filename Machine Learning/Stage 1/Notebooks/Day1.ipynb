### Demander un nombre et afficher :
    • "positif"
    • "négatif"
    • "nul"

```
nb = int(input("Entrer un nombre"))
if nb < 0:
    print("Négatif")
elif nb > 0:
     print("Positif")
else:
     print("nul")
```

## Exercice 4 :
### Demander :
    • âge
    • statut étudiant (yes / no)
    Afficher si la personne a droit à une réduction

```
age = int(input("Quel est votre âge?"))
etudiant = input("Êtes-vous étudiant? (Yes / No)")
if age >= 18 and etudiant == "Yes":
    print("Vous avez le droit à une réduction")
elif etudiant == "No":
    print("Vous n'avez droit à aucune aide")
```
