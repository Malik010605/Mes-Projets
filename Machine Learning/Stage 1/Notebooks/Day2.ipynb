# Jour 2 : Boucles & raisonnement algorithmique
## 1) Boucle `for` :

```python
for i in range(5):
    print(i)
# Affiche 0 1 2 3 4 (en gros affiche de i à n-1)
```

## 2) Méthode `range()` :

```python
list(range(0,16,5)) # Affiche d'un début i jusqu'à la fin n-1 par pas de k
                    # L'affichage se fait grace au "list()"
```

## 3) Boucle `while` :

```python
condition = "condition"
instruction = "instruction"
while condition: # En gros : tant que la condition n'a pas été satisfaite, faire l'instruction
    instruction
```

## 4) Compteur (comme en C --> incrémentateur) :

```python
somme = 0 # On initialise le rang de départ (indice k de la somme)
for i in range(1,10): # Pareil qu'une somme en maths (pour i allant de k à n-1)
    somme += i  # On incrémente le compteur
    print(somme) # On affiche chaque étape
```

## 5) Blindage :

```python
# Ici, on blinde : si i vaut 5 alors on sort de la boucle
if i == 5:
    break
# Ici on blinde : si i est un multiple de deux, on continue
if i % 2 == 0:
    continue
```

## 6) Affichage `f-string` :

```python
age = 20
print(f"Tu as {age} ans") # le f permet à Python de lire la valeur entre {} (il lirait juste le texte sans), c'est moins lourd que d'écrire à chaque fois                                  int(age)
```

## Exercice 2 : Somme
### Calculer la somme de 1 à N (N donné par l'utilisateur)

```python
N = int(input("Entrez l'indice de fin de la somme : "))
somme = 0
for i in range (1,N+1):
    somme += i
print(somme)
```

## Exercice 5 : `while`
### Redemander un nombre tant qu'il est négatif

```python
nb = int(input("Entrez un nombre négatif :"))
while nb<0:
    nb = int(input("Entrez un nombre négatif, pas positif ou nul!"))
print("Bon nombre : ",nb)
```

## Mini projet :
### Créer un jeu qui :
    1) Génère un nombre aléatoire entre 1 et 100
    2) Demande à l'utilisateur de le deviner
    3) Indique :
    • trop grand
    • trop petit
    4) Compte les tentatives
    5) S'arrête quand le nombre est trouvé

```python
import random
nbal = random.randint(1,100)
compteur = 0
nbutil = int(input("Donner le nombre que vous pensez être bon : "))
while nbal != nbutil:
    compteur += 1
    if nbal > nbutil:
        nbutil = int(input("Trop petit! Réessayez :"))
    elif nbal < nbutil:
        nbutil = int(input("Trop grand! Réessayez :"))
compteur +=1
print(f"Bravo! Vous avez trouvé le bon nombre : {nbal} en {compteur} essais")
```
