## 3) Accéder aux données

```python
df = pd.DataFrame({
    "nom" : ["malik", "chamyl", "bob"], # Ici, bob est en indice 2
    "notes" : [10,15,20] # Ici, c'est la note 20 qui est en indice 2 --> Bob a eu une note de 20
})

print(df[["nom","notes"]])
df.loc[2]
```

## 4) Filtrer les données

```python
df = pd.DataFrame({
    "nom" : ["malik", "chamyl", "bob","coco"], # Ici, bob est en indice 2
    "notes" : [10,15,20,8] # Ici, c'est la note 20 qui est en indice 2 --> Bob a eu une note de 20
})

admis = df[df["notes"] >= 10]
refus = df[df["notes"]<10]
print(f"Les admis sont : {admis}")
print(f"Les refusés sont : {refus}")
```

## 5) Les opérations courantes

```python
df = pd.DataFrame({
    "nom" : ["malik", "chamyl", "bob"], # Ici, bob est en indice 2
    "notes" : [10,15,20] # Ici, c'est la note 20 qui est en indice 2 --> Bob a eu une note de 20
})

df["admis"] = df["notes"] >= 10

# Pareil qu'avec Numpy
print(df["notes"].mean())
print(df["notes"].median())
print(df["notes"].std())
print(df["admis"])
print(df)
```

## 6) Nettoyage de datasets

```python
df = pd.DataFrame({
    "nom" : ["malik", "chamyl", "bob"], # Ici, bob est en indice 2
    "notes" : [0,15,20] # Ici, c'est la note 20 qui est en indice 2 --> Bob a eu une note de 20
})
df["admis"] = df["notes"] >= 10

df.isnull() # Détecte une valeur manquante
df.isnull().sum() # Donne le nombre de valeurs manquantes dans la bdd

df.dropna() # Supprime une valeur

df.fillna() # Remplace une valeur par une autre

df.drop_duplicates() # Supprime tous les doublons

print(df)
```

## 7) Agrégation (`groupby`)

```python
df = pd.DataFrame({
    "nom" : ["malik", "chamyl", "bob"], # Ici, bob est en indice 2
    "notes" : [0,15,20] # Ici, c'est la note 20 qui est en indice 2 --> Bob a eu une note de 20
})
df["admis"] = df["notes"] >= 10

print(df.groupby("admis")["notes"].mean())
print(df)
```
