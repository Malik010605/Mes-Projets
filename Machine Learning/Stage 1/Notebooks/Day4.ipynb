# Jour 4 : Structures de données Python
## 1) Les listes (`list`)

```python
notes = [12, 15, 9, 18] # C'est une liste de notes entières

notes[0] # c'est le premier élément de la liste
notes[-1] # c'est le dernier élément de la liste

notes.append(20) # Ajout de la note 14 à la fin de la liste notes
notes.append(2) # Ajout de la note 2 à la fin de la liste notes
notes[1] = 16 # Remplace la deuxième note de la liste (15) par 16
notes[1]

for note in notes: # Boucle qui affiche l'ensemble des notes de la liste
    print(note)
```

## 2) Les tuples (`tuple`)

```python
point = (1,2,3,4,5) # C'est comme une liste sauf qu'on peut pas la modifier
x , y = point # Ici, on décompose la tuple (utile pour les coordonnées ...)
```

## 3) Les dictionnaires (`dict`)

```python
etudiant = {  # C'est un dictionnaire sur un etudiant (associe une clé à une valeur (comme une hashmap en JAVA))
    "nom" : "Malik",
    "age" : 20,
    "notes" : [14,16,17]
}
etudiant["age"] # Accède à la valeur 'age' du dictionnaire etudiant
etudiant["age"] = 21 # Modifie la valeur 'age' dans etudiant
print(etudiant["nom"]) # Affiche une valeur de etudiant
```

## 3) Les ensembles (`set`)

```python
villes = {"Paris", "Toulouse", "Paris", "Marseille", "Paris"} # Exactement pareil qu'un hashset en JAVA, supprime les doublons (ici "Paris")
print(villes)
```

## 4) Structures imbriquées

```python
etudiants = [
    {"nom" : "Malik", "age" : 20, "notes" : [14,15,16]},         # C'est un set de dictionnaires
    {"nom" : "Ismet", "age" : 60, "notes" : [16,20,20]}
]

for e in etudiants:
    moyenne = sum(e["notes"]) / len(e["notes"])
    print(e["nom"], moyenne)
```

## Exercice 3 : Fréquences (`dict`)    **_A revoir car compliqué et méthode intéressante_**
### A partir de : `mots = ["chat", "chien", "chat", "oiseau", "chien", "chat"]`,
### construire un dictionnaire `freq` qui donne le nombre d'occurences de chaque mot


```python
mots = ["chat", "chien", "chat", "oiseau", "chien", "chat"]

frequences = {}

for mot in mots:
    if mot in frequences:
        frequences[mot] += 1
    else:
        frequences[mot] = 1

print(frequences)
```

## Exercice 4 : Déduplication (`set`)
### A partir de villes = ["Paris", "Lyon", "Paris", "Marseille", "Lyon"]
### 1) Créer un `set` pour supprimer les doublons
### 2) Reconvertir en `list`
### 3) Afficher la liste finale

```python
villes = ["Paris", "Lyon", "Paris", "Marseille", "Lyon"]
villes_set = {"Paris", "Lyon", "Paris", "Marseille", "Lyon"}

print(villes_set)

for ville in villes_set:
    if ville in villes:
        villes.remove(ville)
    print(ville)



```
