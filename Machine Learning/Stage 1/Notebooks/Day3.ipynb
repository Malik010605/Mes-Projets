## Mini projet
### Reprendre le projet du Day2 en ajoutant les fonctions suivantes : `generer_nombre()`, `demander_nombre`, `comparer(secret,position)` et `jeu()`

```python
import random

def generer_nombre():
    nbal = random.randint(1,100)
    return nbal

def demander_nombre():
    nbutil = int(input("Donner le nombre que vous pensez être bon : "))
    return nbutil

def comparer(nbal, nbutil):
    compteur = 0
    while nbal != nbutil:
        compteur += 1
        if nbal > nbutil:
            nbutil = int(input("Trop petit! Réessayez :"))
        elif nbal < nbutil:
            nbutil = int(input("Trop grand! Réessayez :"))
    compteur +=1
    print(f"Bravo! Vous avez trouvé le bon nombre : {nbal} en {compteur} essais")

def jeu():
    nbal = generer_nombre()
    nbutil = demander_nombre()
    comparer(nbal, nbutil)

jeu()
```

## Exo perso :
### Coder fonction qui trouve les racines d'un polynome du scd degré (à delta >= 0)

```python
from math import sqrt
def rac_pol2(a,b,c):
    delta = b**2 - 4*a*c
    if delta < 0:
        print("Racines complexes ...")
    elif delta == 0:
        x0 = -b/(2*a)
        print(f"Racine double : {x0}")
    else: 
        x1 = (-b - sqrt(delta))/2*a
        x2 = (-b + sqrt(delta))/2*a
        print(f"Racines : {x1} et {x2}")
rac_pol2(1,-1,-1)
```
